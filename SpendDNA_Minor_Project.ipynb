# SpendDNA – Personal Spending Analysis

## Project Overview

SpendDNA is a personal finance analytics project that analyzes transaction data
to understand spending patterns, vendors, categories, monthly trends,
time-of-day behavior, anomalies, and spending archetypes.

## Objectives

- Clean and standardize transaction data
- Extract and standardize vendors
- Categorize transactions
- Calculate spending statistics
- Analyze monthly spending trends
- Analyze time-of-day spending patterns
- Detect unusual transactions
- Identify spending archetypes

```python
def extract_vendor(description):

    text = str(description).upper()

    if "SWIGGY" in text or "BUNDL" in text:
        return "Swiggy"

    elif "ZOMATO" in text:
        return "Zomato"

    elif "AMAZON" in text or "AMZN" in text:
        return "Amazon"

    elif "ZEPTO" in text:
        return "Zepto"

    elif "ZERODHA" in text:
        return "Zerodha"

    elif "UBER" in text:
      return "uber"

    elif "OLA" in text:
      return "ola"

    elif "NETFLIX" in text:
      return "netflix"

    else:
      return "other"
```

```python
df["Vendor"] = df["Description"].apply(extract_vendor)
```

```python
df["Vendor"].value_counts()
```

```python
category_mapping = {
    "Swiggy": "Food Delivery",
    "Zomato": "Food Delivery",
    "Zepto": "Quick Commerce",
    "Amazon": "E-commerce",
    "Zerodha": "Investments",
    "Uber": "Transport",
    "Ola": "Transport",
    "Netflix": "Subscriptions"
}
```

```python
df["Category"] = df["Vendor"].map(category_mapping)
```

```python
df["Category"] = df["Category"].fillna("Other")
```

```python
df["Category"].value_counts()
```

```python
total_credits = df.loc[
    df["Type"] == "credit",
    "Amount"
].sum()

total_credits
```

```python
total_debits = df.loc[
    df["Type"] == "debit",
    "Amount"
].sum()

total_debits
```

```python
net_change = total_credits - total_debits
```

```python
import numpy as np # Added import for numpy
if total_credits != 0:
    savings_rate = (
        (total_credits - total_debits)
        / total_credits
    ) * 100
else:
    savings_rate = np.nan
```

```python
print("Total Credits:", total_credits)
print("Total Debits:", total_debits)
print("Net Change:", net_change)
print("Savings Rate:", savings_rate)
```

```python
debit_df = df[df["Type"] == "debit"]
```

```python
top_categories = (
    debit_df.groupby("Category")["Amount"]
    .sum()
    .sort_values(ascending=False)
)
```

```python
top_categories.head(10)
```

```python
top_vendors = (
    debit_df.groupby("Vendor")["Amount"]
    .sum()
    .sort_values(ascending=False)
)
```

```python
top_vendors.head(10)
```

```python
monthly_spending = (
    debit_df.groupby("Month_Name")["Amount"]
    .sum()
)
```

```python
month_order = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"
]
```

```python
hourly_spending = (
    debit_df.groupby("Hour")["Amount"]
    .sum()
    .sort_values(ascending=False)
)
```

```python
hourly_spending
```

```python
hourly_transactions = (
    debit_df.groupby("Hour")
    .size()
    .sort_values(ascending=False)
)
```

```python
food_delivery = debit_df[
    debit_df["Category"] == "Food Delivery"
]
```

```python
late_night_food = food_delivery[
    (food_delivery["Hour"] >= 21) |
    (food_delivery["Hour"] <= 1)
]
```

```python
if len(food_delivery) > 0:
    late_night_percentage = (
        len(late_night_food)
        / len(food_delivery)
    ) * 100
else:
    late_night_percentage = np.nan
```

```python
print(
    "Late-night Food Delivery:",
    round(late_night_percentage, 2),
    "%"
)
```

```python
category_mean = (
    debit_df.groupby("Category")["Amount"]
    .transform("mean")
)
```

```python
category_std = (
    debit_df.groupby("Category")["Amount"]
    .transform("std")
)
```

```python
debit_df = debit_df.copy()

debit_df["Z_Score"] = (
    (debit_df["Amount"] - category_mean)
    / category_std
)
```

```python
anomalies = debit_df[
    debit_df["Z_Score"] > 2
]
```

```python
anomalies = anomalies.sort_values(
    "Z_Score",
    ascending=False
)
```

```python
anomalies[
    ["Date", "Description", "Vendor",
     "Category", "Amount", "Z_Score"]
].head(10)
```

```python
category_percentage = (
    top_categories / total_debits
) * 100
```

```python
food_percentage = category_percentage.get(
    "Food Delivery",
    0
)
```

```python
investment_percentage = category_percentage.get(
    "Investments",
    0
)
```

```python
archetypes = []
```

```python
if food_percentage > 15:
    archetypes.append("THE FOODIE")
```

```python
if investment_percentage > 15:
    archetypes.append("THE INVESTOR")
```

```python
ecommerce_percentage = category_percentage.get(
    "E-commerce",
    0
)

if ecommerce_percentage > 15:
    archetypes.append("THE SHOPAHOLIC")
```

```python
if late_night_percentage > 50:
    archetypes.append("THE LATE-NIGHT SNACKER")
```

```python
if savings_rate < 10:
    archetypes.append("THE YOLO SPENDER")
```

```python
print("Spending Archetypes:")

for archetype in archetypes:
    print("-", archetype)
```
